```python
x=df.drop("Class",axis=1)
y=df["Class"]
```

```python
# x.head()
df["Class"].unique()
```

```python
from sklearn.preprocessing import StandardScaler , LabelEncoder 

le = LabelEncoder()
y = le.fit_transform(y)
```

```python
y
```

```python
from sklearn.model_selection import train_test_split 

x_train,x_test,y_train,y_test=train_test_split (
    x,y,test_size=0.2,random_state=42
)
```

```python
# x_train
# x_test
```

```python
scaler = StandardScaler()

x_train_scaled = scaler.fit_transform(x_train)
x_test_scaled=scaler.transform(x_test)
```

ANN

```python
import torch
import torch.nn as nn
import torch.optim as optim 
from torch.utils.data import DataLoader , TensorDataset 
```

```python
x_train_tensor = torch.tensor(x_train_scaled , dtype = torch.float32)
y_train_tensor = torch.tensor(y_train , dtype = torch.long)#long because we perform multiclass classifiction  
                                                           # so there loss fununction(cross entropy loss) are accept targrt value in long dataset 

x_test_tensor = torch.tensor(x_test_scaled , dtype = torch.float32)
y_test_tensor = torch.tensor(y_test , dtype = torch.long)
```

```python
# x_test_tensor.size()
```

```python
# y_train_tensor.size()
```

```python
train_dataset = TensorDataset(x_train_tensor, y_train_tensor)
test_dataset = TensorDataset(x_test_tensor, y_test_tensor)
```

```python
train_loader = DataLoader(train_dataset, batch_size=32, shuffle=True)
test_loader = DataLoader(test_dataset, batch_size=32)
```

```python
#build our Model
```

```python
class ANN(nn.Module):
    def __init__(self):
        super(ANN, self).__init__()

        self.model = nn.Sequential(
            nn.Linear(x.shape[1],64),
            nn.ReLU(),

            nn.Linear(64,64),
            nn.ReLU(),

            nn.Linear(64,7)
        )

    def forward(self,x):
        return self.model(x)
```

```python
model = ANN()

#loss & optimaizer

criteria = nn.CrossEntropyLoss()
optimizer = optim.Adam(model.parameters())
```

```python
# Training The ANN

epochs = 100
for epoch in range(epochs):
    model.train()

    running_loss = 0.0

    for xb, yb in train_loader:
        optimizer.zero_grad()
        
        output = model(xb)# forward propogation 
        loss = criteria(output,yb)# loss calculate 
        loss.backward() #backward propogation 
        optimizer.step()# parameters update 

        running_loss += loss

    training_loss = running_loss / len(train_loader)

    print(f"epoch = {epoch}/{epochs}, loss = {training_loss}")
        
```

```python
# Evaluate
model.eval()#parameter are fixed. no change through training data 

total = 0
correct = 0

with torch.no_grad():#no gardient compute 
    for xb,yb in test_loader:
        output = model(xb) #[0.2,0.5,1.3,-0.5,..] 7 values
        _,predicted = torch.max(output,1)#torch.max can give max value and index number and 1 is direction and access and also _ it is use because we have no use of max value 

        correct += (predicted == yb).sum().item()
        total += yb.size(0) #actual sample in each btach #we are use size because any number so i can give me reaminder and this remainder not same size of batch .

print("Total vals: ",total)
print("correct vals: ", correct)

print("Accuracy: ", correct/total*100)
```
